```python
import pandas as pd
import yfinance as yf
```

## Ekstraksi dan Prapemrosesan Data (Data Preprocessing)
Dataset historis saham BBCA dimuat ke dalam struktur *DataFrame*. Tahapan ini mencakup penyesuaian penamaan kolom, penetapan *DatetimeIndex*, dan reduksi dimensi pada fitur yang tidak memiliki nilai informasi prediktif.

```python
def load_and_clean_data(ticker_symbol: str = "BBCA.JK", period: str = "5y") -> pd.DataFrame:
    print(f"Mengunduh data {ticker_symbol} dari Yahoo Finance...")
    
    ticker = yf.Ticker(ticker_symbol)
    df = ticker.history(period=period)
    
    if df.index.tz is not None:
        df.index = df.index.tz_localize(None)
        
    df.drop(columns=['Dividends', 'Stock Splits'], errors='ignore', inplace=True)
    
    print(f"Data berhasil diproses. Dimensi awal dataset: {df.shape}")
    return df

bbca = load_and_clean_data("BBCA.JK", period="20y")
bbca.tail(5)

bbca.to_csv("bbca_20years.csv")
```
